## Catching Errors

### Very useful to have a way to "catch errors" and keep a program running.

```python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("You tried to divide by zero!")
```

```python
print("Start")

try:
    x = 10 / 0
    print("This line will NOT run")
except ZeroDivisionError:
    print("Oops — divided by zero!")

print("Program keeps running here")
```

### Note that you after catching the exception, you do NOT return to the try block !!!

```python
try:
    number = int("hello")
except ValueError:
    print("That is not a valid integer!")
```

```python
try:
    with open("does_not_exist.txt", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("File not found. Did you spell the filename correctly?")
```

```python
try:
    with open("missing.txt") as f:
        data = f.read()
except Exception as e:
    print("Error type:", type(e).__name__)    # Can be useful. We won't use much.
    print("Error message:", e)                # Very useful.
```
