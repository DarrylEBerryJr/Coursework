### 1.) Resonance curve for a lightly damped, driven oscillator

You have already determined the amplitude of an oscillator driven by an external torque or force oscillating at   
its own frequency $\omega$. Here the idea is to vary the driver's $\omega$ and see how the amplitude $A$ of the    
oscillator changes. So, you want to compute $A(\omega)$. Ultimately, you will **plot** the quantity $A(\omega/\omega_0)$   
so you can easily see how the oscillator's amplituide changes when you excite it near its 'natural' frequency $\omega_0$.   
By definition, the amplitude $A$ is positive even though the displacement of an oscillator is both positive and negative.  
This resonance phenomenon is quite common in nature.

The differential equation for our one-dimensional driven oscillator can be written as

$$
x'' + 2 \gamma x' + \omega_0^2 x = (F_0/m) \cos(\omega\,t)
$$

Here, the quantity $x(t)$ is the displacement of whatever is oscillating from its equilibrium position as a function of time $t$ and       
 $\gamma$ is a parameter that determines how much frictional damping there is acting on the oscillator   
whose mass is $m$. The quantity $\omega_0$ is the so-called natural frequency of the oscillator,   
the (angular) frequency the oscillator oscillates at if there were no friction and no driver.   
The oscillatory force applied by the the driver has a magnitude of $F_0$ and oscillates with a frequency $\omega$.   
Don't confuse $\omega$ with $\omega_0$. They refer to different quantities.


For simplicity, we can set $m = \omega_0 = F_0 = 1$. (The units will work out.) Our oscillator is    
'lightly damped' so we can start with $\gamma = 0.1$. We can always change this later if we want.


Determine  the shape of the curve showing the steady-state amplitude of oscillation as a function      
of the oscillation frequency $\omega$ of the 'driver.' To make things easier to understand, plot $A(\omega/ \omega_0)$.

#### Guidance

You want to solve the ODE for a **range** of driving frequencies $\omega$ around the natural frequency $\omega_0$. That means you   
need to solve the ODE multiple times, once for each choice of driving frequency. You can scan frequencies $\omega$ from, say, $0.4\omega_0$ to $1.6 \omega_0$. Furthermore, for each choice of driving frequency you only want the steady state value of the dependent variable $x$. That means you should ignore the early time behavior of the solution. One approach is to solve the ODE for, say, a total time equal to 100 periods of your driver frequency $\omega$ and then keep only the last 30 periods or so of the solution. You have already seen from the driven simple pendulum that the motion of the pendulum was very complicated at first but then settled down to steady state behavior. We want the steady state behavior of our oscillator. You can start the oscillator at $x=0$ and $x'=0$.

If you are successful, your plot of steady state $A$ versus $\omega/\omega_0$ will look like a "peaky" curve that diminishes as you move in either direction away from 1.
