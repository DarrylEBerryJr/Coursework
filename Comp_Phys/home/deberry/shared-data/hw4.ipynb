#### Introduction

Consider a simple pendulum made from a rigid rod of length $L$ and a mass $m$ attached at the end of the rod.   
The rod has a negligible mass compared to $m$. Additionally, the rod is suspended from a knob that can exert    
a torque on the end of the rod away from the hanging mass that can aid or resist the normal oscillation of the    
pendulum. The displacement angle $\theta$ of the mass is measured with respect to the vertical, like any other    
pendulum you have seen. A positive displacement moves $\theta$ in a counterclockwise direction when viewed    
perpendicularly to the plane of free oscillation.

The second order ODE for this system in its full glory is
$$
    \ddot\theta +\frac{c}{mL^2}\dot\theta  + \frac{g}{L}\sin\theta = \frac{A}{mL^2}\cos\omega_d t
$$


Yikes! What does all this stuff mean? Well, $c$ is a parameter that governs how much friction the swinging    
pendulum experiences. The larger $c$ is, the great the friction. The second term from the left is the "damping"    
term and controls the rate at which oscillations tend to die out. The symbol $g$ is the acceleration due to gravity.    
The parameter $A$ is the torque applied to the top end of the rod to aid or resist oscillation in the oscillation plane.    
The torque is applied repetitively with an angular frequency $\omega_d$. The term on the right is often called    
the "driving force" even though it does not strictly speaking represent a force.  


You are used to seeing this equation in intro physics but for the simple case $c=A=0$ and for $\sin\theta$ approximated    
as $\theta$, appropriate when $\theta < 10^{\circ}$ or so. This situation occurs when gravity alone acts on the pendulum,   
there is no damping from friction, and the amplitude of oscillation is not too large.

The above equation is a wee bit complicated for our purposes so we will simplify it for solving by combining   
parameters. We rewrite it as
$$
\ddot\theta + \gamma \dot\theta + \frac{g}{L}\sin\theta = F_d\cos\omega_d t
$$
