## Representation of 0.1 in binary (float64)

### Step 1: Convert 0.1 to binary

Take the fractional part and repeatedly multiply by 2, keeping track of the integer part each time:

0.1 × 2 = 0.2 $\rightarrow$ integer part = 0

0.2 × 2 = 0.4 $\rightarrow$  integer part = 0

0.4 × 2 = 0.8 $\rightarrow$ integer part = 0

0.8 × 2 = 1.6 $\rightarrow$ integer part = 1 (keep 1, continue with 0.6)

0.6 × 2 = 1.2 $\rightarrow$ integer part = 1 (keep 1, continue with 0.2)

0.2 × 2 = 0.4 $\rightarrow$ integer part = 0

0.4 × 2 = 0.8 $\rightarrow$ integer part = 0

0.8 × 2 = 1.6 $\rightarrow$ integer part = 1

$\dots$ and the pattern repeats.

So in binary:

0.1(base 10) = 0.0001100110011001100110011…(base 2)

It'a a repeating binary fractiom with the block 0011 repearing forever - just like 1/3 = 0.333 $\dots$
repeats in decimal.


### Step 2: IEEE-754 storage

Since computers only have finite bits, this repeating binary fraction must be truncated/rounded.

**In 32-bit float (float32):**

0.1 $\approx$ 0.10000000149011612


**In 64-bit float (float64):**

0.1 $\approx$ 0.10000000000000000555

This is why:
0.1 + 0.2 == 0.3 is False in Python code



**IEEE-754 double layout**

A 64-bit float is divided into three fields:

$$
\begin{array}{lcl}
\textrm{Field}      & \textrm{Bits} & \textrm{Purpose} \\
\textrm{Sign}       & 1             & 0 = \textrm{positive}, 1 = \textrm{negative} \\
\textrm{Exponent}   &  11           & \textrm{Encodes the power of 2 with a bias of 1023}\\
\textrm{mantissa (fraction)}  & 52  & \textrm{Encodes the fractional bits after an implicit leading 1}
\end{array}
$$


**Normalize**

IEEE floats always normalize numbers to the form

$(-1)^{\textrm{sign bit}}\times1.\textrm{fraction bits}\times 2^e$


For 0.1:

0.1 (base 10) = 1.1001100110011$\dots$(base 2) $\times 2^{-4}$


So:

Sign = 0 (positive)

Exponent = $−4$


Stored exponent = $−4 + 1023 = 1019$
$−4 + 1023 = 1019 \rightarrow$ 01111111011 (binary)


Mantissa = the bits after the leading 1. in 1.100110011…
