
# Plotting with Matplotlib — Essentials

This notebook introduces the basics of plotting with **matplotlib.pyplot**.  
You will learn to create simple plots, label axes, add titles and legends, and make histograms and scatter plots.

At the end there are exercises to practice.


```python
import sys, site, importlib, numpy as np
```

```python
import matplotlib
```

## 1. Importing matplotlib

```python

import matplotlib.pyplot as plt
import numpy as np

# Create some sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

## 2. Line Plot

```python

plt.plot(x, y)
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

## 3. Scatter Plot

```python

np.random.seed(0)
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color="red", marker="o")
plt.title("Scatter Plot Example")
plt.xlabel("x values")
plt.ylabel("y values")
plt.show()
```

## 4. Histogram

```python

data = np.random.randn(1000)  # normal distribution
plt.hist(data, bins=30, color="skyblue", edgecolor="black")
plt.title("Histogram Example")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## 5. Multiple Plots on Same Axes

```python

x = np.linspace(0, 2*np.pi, 100)
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.title("Sine and Cosine")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()
plt.show()
```

## 6. Subplots

```python

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10,4))

ax1.plot(x, y1)
ax1.set_title("sin(x)")

ax2.plot(x, y2)
ax2.set_title("cos(x)")

plt.show()
```
