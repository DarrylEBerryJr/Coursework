## File I/O in Python
Working with Data Files

In computational physics, you will constantly read experimental data from text files and write output from simulations. Python makes this straightforward. This notebook demonstrates the most common patterns: reading plain text, CSV, and writing your own data files.

### 1. Reading an Entire File as a Single String
The built‑in `open()` function is the standard way to read files.

```python
import numpy as np
```

```python
# Reads a whole file into one string
with open('example.txt', 'r') as f:
    text = f.read()
print(text)
```

### 2. Reading Line‑by‑Line
Useful when files are large or structured line‑wise.

```python
### Iterate over lines
data_file = 'dow.txt'
with open(data_file, 'r') as f:
    for line in f:
        print(line.strip())
```

### 3. Reading Numerical Data with NumPy
Data files often contain columns of numbers. NumPy offers fast one‑liners for these.

```python
import numpy as np
data_file = 'muon.data'
data = np.loadtxt(data_file, float)
print(data.shape)   # what does the object data "look" like?
print(data[:5])  # first 5 rows
print(data[:5,0])  # print just the first value of each row
```

If the file has multiple columns, NumPy loads it into a 2D array. You can unpack columns like this:

```python
data_filoe = 'muon.data'
x, y = np.loadtxt('muon.data', float, unpack=True)
print(x[:5], y[:5])
```

### 4. Plotting Data After Loading It
Read a file and immediately plot it.

```python
import matplotlib.pyplot as plt
import numpy as np
data_file = 'muon.data'
x, y = np.loadtxt(data_file, int, unpack=True)   # data are integers
mask = x < 40000         # ignore data lines that correspond to 'timed out' electronics
x_good = x[mask]/1000   # convert from ns --> microsec
#                         This also works: x_good = x[x < 40000]/1000
plt.hist(x_good, bins=50, color='orange', edgecolor='black')  
plt.xlim(0., 20.)       # 20 is chosen somewhat arbitrarily
plt.xlabel('Decay time [$\mu\,s$]')
plt.ylabel(r'$\mu$ decays')
plt.title(f'Data from {data_file}')
plt.grid(True)
plt.show()
print('The number of muons collected =', len(x_good))
delta_t = y[-1] - y[0]
print(f'Collected over {delta_t} seconds or {delta_t/86400:.2f} days')
```

### 5. Writing Data to a File
You typically save your simulation results to a file for later analysis.

```python
# Write your own dataset
output_file = 'sine_output.txt'
x = np.linspace(0, 2*np.pi, 50)
y = np.sin(x)
with open(output_file, 'w') as f:
    for xi, yi in zip(x, y):    # x and y have same length, zip walks through the arrays in order and makes pairs.
        f.write(f'{xi} {yi}\n')    # the \n character puts the pairs in a separate line
```

### Using NumPy for Writing
NumPy can also write arrays directly.

```python
xy = np.column_stack((x, y))    # takes the same-size 1-D arrays x and y, and treats each as a column.
np.savetxt('sine_output_np.txt', xy, fmt='%.6f')
```

### 6. Handling Files with Headers, Comments, or Skipped Lines
Real experimental files often include comments or metadata (data about the data). NumPy can handle this.

```python
input_data = 'dow.data'
data = np.loadtxt(input_data, comments='#', skiprows=2)
print(data[:5])    # print first 5 rows
```

**What if** the numbers on each line are separated by commas, a so-called CSV file? Numpy can handle this.

```python
input_data = 'trajectory_output.csv'
data = np.loadtxt(input_data, comments='#', delimiter=',', skiprows=1)
print(data[:5])    # print first 5 rows
```
