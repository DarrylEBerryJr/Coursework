
# Importing Libraries and Functions in Python



## 1. Import a whole library

```python

import math

# Now we can use math functions with the prefix "math."
print("Square root of 16 is", math.sqrt(16))
print("Pi is", math.pi)
print("Rounded Pi is:", round(math.pi,3))
```

## 2. Import with an alias

```python

import numpy as np

# Use "np" instead of typing "numpy" each time
arr = np.array([1, 2, 3, 4, 5, 6])
print("Numpy array:", arr)
print("Mean:", np.mean(arr))
```

## 3. Import only specific functions

```python

from math import sqrt, cos

print("Square root of 36 is", sqrt(36))  # Note: no need for prefix np on sqrt
print("Cosine of 0 is", cos(0))
```

## 4. Import everything (not recommended)

```python

from math import *

print("Factorial of 6 is", factorial(6))
```


---
## Summary
- `import module` $\rightarrow$ import the whole library.  
- `import module as alias` $\rightarrow$ shorter name for convenience.  
- `from module import func1, func2`$\dots$ $\rightarrow$ import only what you need.  
- `from module import *` $\rightarrow$ imports everything (can cause name conflicts, avoid in practice).
