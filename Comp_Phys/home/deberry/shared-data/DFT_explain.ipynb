### Some Math Behind the Magic: Computing $X_k$ in the DFT 

##### Brief Overview

You have seen the comment that trhe FFT is a fast method to compute A DFT. Fine, but how do you compute   
the all important quantity $X_k$? Let's see how to do it for the straightforward case of the brute force DFT.

A discrete Fourier transform (DFT) rewrites or approximates  a signal $x(t)$ as a sum of sinusoids:

$$
x(t)\approx \Sigma_{k=0}^{N-1}X_k e^{i2\pi f_kt},
$$   
where we use the famous identity $e^{i\theta} = \cos\theta + i\sin{\theta}$ and $f_k$ is a frequency.   

Each $X_k$ is a complex number, containing   

* A **magnitude:** telling you how strong that frequency component is
* A **phase:** telling you where the wave starts in its cycle

We usually focus on the magnitude $|X_k|$, the amplitude of the sinusoid at frequency $f_k$.



#### Example: 50 Hz sine wave

Suppose
$$
x(t) = A sin(2\pi 50 t)
$$

then the DFT would give a big magnitude at 50 Hz and everyhing else $\sim 0$ (except for numerical noise):   
$$
∣X_{50}∣\approx \frac{AN}{2}
$$

The factor $N$ (number of data points) is not needed now so we divide ∣X_{50}∣ by $N$. You have already seen this. To get our amplitude, one of the things we care about, we need to multiply ∣X_{50}∣ by 2. Overall, this is why you see for the FFT we execute:  

`X_mag = 2 * np.abs(X[:N//2]) / N`

But where precisely do these factors come from? And why do we only use the first half of the $X$ array by using the `//` operator? 




A real sine always produces two equal-magnitude DFT peaks:

* one at +f
* one at −f (the “mirror”)

Their bin indices are:   

* positive-frequency bin: $k_0=\frac{f_0N}{f_s}$
* negative-frequency mirror: $N - k_0$

This holds for **any** $N$.

\end{array}
$$


#### Frequency resolution

From Fourier analysis of a signal measured for a **finite** time, the final result involves the multiplication of the Fourier transform of a rectangular window whose length equals the time taken by the measurement. (You never saw this effect in Physics classes because you were always integrating over an infinite amount of time. Look back at your classical mechanics text.) The shape of the Fourier transform of a rectangular-looking function of width $T$ looks like a "sinc" function $\frac{\sin \pi f T}{\pi f T}$. This looks like a central lobe with much smaller wiggles on either side of it. 

The main lobe of $\text sinc(x)$ has a width $\Delta f = \frac{1}{T}$.  




Frequency resolution is determined by **total observation time**, not sample spacing.
