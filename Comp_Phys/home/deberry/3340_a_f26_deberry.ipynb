```python
#default opening
import numpy as np   
#import is a keyword, making library available
#numpy = numerical py, pre-existing code
#as allows alias for imported code
#np is common abbreviation for numpy, utilize current abbreviations for ease
from scipy.optimize import bisect
#from identifies library will pick a portion thereof
#import after 'from' picks a portion of the afformentioned library
import matplotlib.pyplot as plt
#allows making plots
```

```python


def f(x):
#defines a function named f taking input variable x
    return np.sqrt(x) + np.sqrt(2) - 2
    #returns a value that is square root in numpy of variable + sqrt 2 minus 2

root = bisect(f, 0, 1)
#find root of function with bisect and name the result 'root'
#bisect - finds root of function f, on the interval [0,1] with the 1 being the OPPOSITE sign of the value of f
#see class notes 2026-08-25 for detailed notes on bisection process with graph
print(root)
```

```python
import numpy as np
from scipy.optimize import bisect
import matplotlib.pyplot as plt

def f1(x):
    return np.e**(np.sqrt(x)) - 2
#Finding x of e^sqrt(x) = 2, equivalent to finding root of f(x) = e^sqrt(x) - 2
root = bisect(f1, 0, 2)

print(root)
print(f"{root:.3f}") #formatted statement to three decimals of number
#all the folling in print()
#f = formatted statement
#"{variable:.#f}" # is # of decimals for the f fixed number of decimals
print(f"{root:.3e}") #formatted statement to three decimals of number
#e exponent instead of f fixed for e notation
print(f"{root:.3g}") #formatted statement to three decimals of number
#g for unknown number of decimal places in answer
print(np.pi)
```
